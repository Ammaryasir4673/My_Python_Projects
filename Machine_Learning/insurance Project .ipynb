**lets create a linear regression model for insurance data**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#first upload the cleared data that we saved in name of insurance_charges. 
```

```python
final_df = pd.read_csv("cleaned insurance.csv")
```

```python
final_df
```

```python
from sklearn.model_selection import train_test_split
```

```python
X=final_df.drop("charges",axis=1) # X contain all features that changes charges columns values
y=final_df["charges"]# only charges column
```

**now we will split data into test train**

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

**Now create and Train the linear Regression model**

```python
from sklearn.linear_model import LinearRegression  #import linearRegression model
model=LinearRegression()
model.fit(X_train,y_train)# model will learn the pattern From xtrain and ytrain
y_pred=model.predict(X_test) # it will predict the y on basis of Xtest data
```

**Now compare the y_pred with y_test by using R square and adjusted R square (Model evaluation)**

```python
from sklearn.metrics import r2_score #import R2
r2=r2_score(y_test,y_pred)
r2
# model is giving 80% accuracy 
```

```python
#now compare via adjusted R2. here we required no. of rows in X_test (n) and no. of colums or features in X_test(p)
```

```python
X_test.shape
```

```python
n=X_test.shape[0] # in shape first or  o index element represents no. of rows and second one represents no. of columns
p=X_test.shape[1]
adjusted_r2=1-((1-r2)*(n-1)/(n-p-1))
adjusted_r2
```
