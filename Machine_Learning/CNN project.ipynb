```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import tensorflow as ft
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Flatten
from tensorflow.keras.layers import Dropout
from tensorflow.keras.utils import to_categorical
```

```python
df=pd.read_csv("mnist_train.csv")
df_test=pd.read_csv("mnist_test.csv")
```

```python
df
```

```python
X_train=df.drop("label",axis=1).values
y_train=df["label"].values
X_test=df.drop("label",axis=1).values
y_test=df["label"].values 
```

```python
#above we need only values so we extract values
```

```python
X_train=X_train.astype("float32")/255 #Normalization converts pixel values int 0 and 1
X_test=X_test.astype("float32")/255
```

```python
X_train_img=X_train.reshape(-1,28,28) #intially image was in 1X785 now we reshapre into 28x28
X_test_img=X_test.reshape(-1,28,28)
```

```python
#now do one hot encodeing for y train and test becaue they are 0 to 9
```

```python
y_train_cat=to_categorical(y_train,10)
y_test_cat=to_categorical(y_test,10)
```

**Perceptron Model**

```python
perceptron=Sequential([
    Flatten(input_shape=(28,28)),
    Dense(10,activation="softmax")
])
```

```python
perceptron.compile(optimizer='sgd',loss="categorical_crossentropy",metrics=["accuracy"])
```

```python
history_percp = perceptron.fit(
    X_train_img,
    y_train_cat,
    epochs=5,
    batch_size=32,
    validation_data=(X_test_img, y_test_cat),
    verbose=1
) #train model
```

```python
acc_percp=perceptron.evaluate(X_test_img,y_test_cat,verbose=0)[1]
acc_percp
```

**ANN Model**

```python
ann=Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128,activation="relu"),
    Dense(64,activation="relu"),
    Dense(10,activation="softmax")
])
```

```python
ann.compile(optimizer='adam',loss="categorical_crossentropy",metrics=["accuracy"])
```

```python
history_ann = ann.fit(
    X_train_img,
    y_train_cat,
    epochs=2,#use only 2 it take long time if it is high
    batch_size=32,
    validation_data=(X_test_img, y_test_cat),
    verbose=1
)
```

```python
acc_ann=ann.evaluate(X_test_img,y_test_cat,verbose=0)[1]
acc_ann
```

**CNN model**

```python
# channel should be for cnn data
```

```python
X_train_cnn=X_train.reshape(-1,28,28,1) # last 1 in bracket show it is grayscale data. if it is 3 then it is rgb.
X_test_cnn=X_test.reshape(-1,28,28,1)
```

```python
from tensorflow.keras.layers import Conv2D
from tensorflow.keras.layers import MaxPooling2D
cnn=Sequential([
    Conv2D(32,kernel_size=(3,3),activation="relu",input_shape=(28,28,1)),
    MaxPooling2D(pool_size=(2,2)),
    Conv2D(64,kernel_size=(3,3),activation="relu",input_shape=(28,28,1)),
    MaxPooling2D(pool_size=(2,2)),
    Flatten(),
    Dense(128,activation="relu"),
    Dropout(0.5),
    Dense(10,activation="softmax")
])
```

```python
cnn.compile(optimizer='adam',loss="categorical_crossentropy",metrics=["accuracy"])
```

```python
history_cnn = cnn.fit(
    X_train_cnn,
    y_train_cat,
    epochs=2,
    batch_size=32,
    validation_data=(X_test_cnn, y_test_cat),
    verbose=1
)
```

```python
acc_cnn=cnn.evaluate(X_test_cnn,y_test_cat,verbose=0)[1]
acc_cnn
```
